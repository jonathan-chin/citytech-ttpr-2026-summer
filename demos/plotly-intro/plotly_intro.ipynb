# Interactive Charts with Plotly

Seaborn makes **static** charts, great for a report. **Plotly** makes **interactive**
charts, great for a dashboard: you can **hover**, **zoom**, **pan**, and toggle series.

This notebook uses the real **2018 Central Park Squirrel Census** (3,023 squirrels).

> **Run this notebook** to feel the difference. On GitHub the charts look static;
> live in Jupyter they are interactive.

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("squirrel_census.csv")
print(df.shape)
df[["unique_squirrel_id", "x", "y", "primary_fur_color", "shift"]].head()
```

## A map of squirrels

Each squirrel has an `x` (longitude) and `y` (latitude). Plot them and the points
trace out **Central Park**. Color by fur color.

**Hover** over a point to see its details. **Drag** to zoom into a corner.

```python
fur = df.dropna(subset=["primary_fur_color"])

fig = px.scatter(
    fur, x="x", y="y", color="primary_fur_color",
    hover_data=["unique_squirrel_id", "shift"],
    labels={"x": "longitude", "y": "latitude", "primary_fur_color": "fur color"},
    title="Central Park squirrels, by fur color",
)
fig.update_traces(marker=dict(size=6, opacity=0.7))
fig
```

## Plotly Express reads like seaborn

The pattern is the same idea as seaborn: pass the DataFrame, name the columns.
`px` is Plotly's high-level API.

```python
acts = ["running", "chasing", "climbing", "eating", "foraging"]
counts = df[acts].sum().sort_values(ascending=False).reset_index()
counts.columns = ["activity", "squirrels"]

fig = px.bar(counts, x="activity", y="squirrels",
             title="What are the squirrels doing?")
fig.show()
```

## The interactivity is the point

Try these on the charts above:

- **Hover** a bar or point for exact values
- **Click and drag** to zoom; **double-click** to reset
- **Click a legend entry** to hide or isolate that group

None of that is possible in a static image.

```python
# color the map by age instead: same call, one changed argument
age = df.dropna(subset=["age"])
age = age[age["age"].isin(["Adult", "Juvenile"])]

fig = px.scatter(
    age, x="x", y="y", color="age",
    labels={"x": "longitude", "y": "latitude"},
    title="Central Park squirrels, by age",
)
fig.update_traces(marker=dict(size=6, opacity=0.7))
fig
```

## Save it as a web page

A Plotly figure can be written to a standalone **HTML file** you can share or
drop into a dashboard. That is what makes it a dashboard tool.

```python
fig.write_html("squirrels.html")
print("wrote squirrels.html (open it in a browser)")
```

## Seaborn or Plotly?

- **Seaborn**: static, publication-quality charts for a **report** or a slide
- **Plotly**: interactive charts for a **dashboard** or exploration

Same data, two jobs. Pick the tool that fits how the chart will be **used**.
