# EDA with pandas: NYC Squirrel Census

Exploring the **2018 Central Park Squirrel Census** (3,023 squirrel sightings)
using pandas' built-in EDA tools.

Data: [2018 Central Park Squirrel Census - Squirrel Data](https://data.cityofnewyork.us/Environment/2018-Central-Park-Squirrel-Census-Squirrel-Data/vfnx-vebw) (NYC Open Data, dataset `vfnx-vebw`).

The usual first-look ritual: **head -> info -> describe -> check missing -> value_counts**.

```python
# Load the CSV sitting next to this notebook into a DataFrame.
import pandas as pd

df = pd.read_csv('squirrel_census.csv')
df
```

## First look: structure

```python
# head(): the first rows (default 5). A quick peek at the columns and values.
df.head()
```

```python
# tail(): the last rows. Handy for checking the end of the data too.
df.tail()
```

```python
# shape: (rows, columns) as a tuple. No parentheses -- it's an attribute.
df.shape
```

```python
# info(): column names, non-null counts, and dtypes.
# The non-null counts are your first hint about missing data.
df.info()
```

```python
# sample(): random rows -- less biased than always looking at the top few.
df.sample(5)
```

## Summary statistics

```python
# describe(): summary stats for the NUMERIC columns
# (count, mean, std, min, quartiles, max).
# Note: `date` is stored as an integer, so its stats are meaningless --
# a good reminder that a column can have the wrong type.
df.describe()
```

```python
# describe(include='all'): every column, including text.
# Text columns get count, unique, top (most common), and freq instead.
df.describe(include='all')
```

```python
# Single-column stats. `hectare_squirrel_number` is how many squirrels were
# counted in a given hectare, so these are actually meaningful.
num = df['hectare_squirrel_number']

print('mean:  ', num.mean())     # average
print('median:', num.median())   # middle value
print('std:   ', num.std())      # spread (standard deviation)
```

```python
# min, max, and quantile (percentiles).
print('min:', num.min())
print('max:', num.max())
print('25th percentile:', num.quantile(0.25))
print('90th percentile:', num.quantile(0.90))
```

## Categorical columns

```python
# value_counts(): how many times each value appears. Great for categories.
# Most Central Park squirrels are Gray. Pass dropna=False to also count NaNs.
df['primary_fur_color'].value_counts(dropna=False)
```

```python
# unique(): the distinct values in a column.
# nunique(): how many distinct values there are (a count).
print('unique fur colors:', df['primary_fur_color'].unique())
print('how many:', df['primary_fur_color'].nunique())
```

## Missing data

```python
# isna() marks each cell True/False for missing; .sum() counts per column.
# Sort to see the columns with the most gaps first.
df.isna().sum().sort_values(ascending=False)
```

```python
# dropna() removes rows with missing values.
# Drop only rows missing a primary_fur_color, then compare the row counts.
before = df.shape[0]
clean = df.dropna(subset=['primary_fur_color'])
after = clean.shape[0]
print(f'rows before: {before}, after dropping missing fur color: {after}')
```
