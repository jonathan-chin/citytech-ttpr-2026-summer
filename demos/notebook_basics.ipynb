# Jupyter Notebook Basics

A quick tour of how notebook cells work. Run a cell with **Shift+Enter**.

Things to notice as you go:

- Cells share one **in-memory state** (as if every variable were global)
- You can run cells **out of order**, which can create confusing results
- Outputs can go **stale** if you change a cell but don't re-run the ones below it
- The notebook **saves outputs inside the .ipynb file** (a .py holds only code)
- When in doubt: **Kernel > Restart Kernel and Run All Cells**

```python
# Setup: import the random module and start with an empty list.
# `numbers` lives in the notebook's shared memory, so every cell below can see it.
import random

numbers = []
numbers
```

```python
# Append a random number (1-100) to the list.
# Run this cell MULTIPLE times and watch the list grow: each run adds one more.
numbers.append(random.randint(1, 100))
numbers
```

```python
# Remove (pop) the last number off the list.
# Run it a few times, or out of order, and the list shrinks.
# Pop more than you've appended and you get an error: that's stale state.
numbers.pop()
numbers
```

```python
# Average of whatever is CURRENTLY in `numbers`.
# The result depends on how many times (and in what order) you ran the
# append/pop cells above: this cell only sees the current in-memory state.
average = sum(numbers) / len(numbers)
average
```

## Try the notebook quirks

1. Run **append** a few times, then run **average**.
2. Run **append** again but do NOT re-run **average**: its output is now **stale**.
3. Run cells **out of order** and watch the results stop making sense.
4. Pop until the list is empty, then run **average**: it errors on the stale state.

For a clean, reproducible run, use **Kernel > Restart Kernel and Run All Cells**.
It clears memory and runs every cell top to bottom, like a script.
